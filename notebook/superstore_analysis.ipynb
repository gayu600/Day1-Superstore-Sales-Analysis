## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

## 5. Feature Engineering

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df.info()
```

## 6. Create New Columns

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month_Name'] = df['Order Date'].dt.month_name()
df['Profit_Margin'] = (df['Profit'] / df['Sales']) * 100
df.info()
```

##  7. Customer Segmentation

```python
# 1. Customer Sales Analysis
```

```python
customer_sales = df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False)

customer_sales.head(10)
```

```python
# Customer Sales Analysis bar chart
top_customers = customer_sales.head(10)

plt.figure(figsize=(10,6))

top_customers.plot(kind='barh')

plt.title("Top 10 Customers by Sales")
plt.xlabel("Sales")
plt.ylabel("Customer")

plt.gca().invert_yaxis()

plt.savefig("../charts/top_customers_sales.png")

plt.show()
```

```python
# 2.Customer Profit Analysis
```

```python
customer_profit = df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False)

customer_profit.head(10)
```

```python
#  top 10 -Customer Profit Analysis 
top_profit_customers = customer_profit.head(10)

plt.figure(figsize=(10,6))

top_profit_customers.plot(kind='barh')

plt.title("Top Customers by Profit")
plt.xlabel("Profit")

plt.gca().invert_yaxis()
plt.savefig("../charts/top_customers_profit.png")
plt.show()
```

```python
#3. Loss-Making Customers
```

```python
loss_customers = customer_profit.sort_values().head(10)

plt.figure(figsize=(10,6))

loss_customers.plot(kind='barh')

plt.title("Top Loss-Making Customers")
plt.xlabel("Profit")

plt.gca().invert_yaxis()
plt.savefig("../charts/top_loss_making_customers.png")
plt.show()
```

```python
# 4. Order Frequency
```

```python
customer_orders = df.groupby('Customer Name')['Order ID'].count().sort_values(ascending=False)

customer_orders.head(10)
```

```python
# 5. Customer Segmentation Logic
```

```python
# Create Total Customer Sales Table
customer_segment = df.groupby('Customer Name')['Sales'].sum().reset_index()
df.info()
```

```python
# 6. View Segments
```

```python
customer_segment.head()
```

```python
customer_segment['Segment'] = pd.qcut(
    customer_segment['Sales'],
    q=4,
    labels=['Low Value', 'Medium Value', 'High Value', 'VIP']
)
```

```python
#7. Count Segment Customers
```

```python
customer_segment['Segment'].value_counts()
```

```python
# 8.Segment Visualization
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Segment',
    data=customer_segment,
    order=['Low Value', 'Medium Value', 'High Value', 'VIP']
)

plt.title("Customer Segmentation")
plt.savefig("../charts/customers_segmentation.png")
plt.show()
```

```python
# 9. Average Sales by Segment
```

```python
segment_avg_sales = customer_segment.groupby('Segment')['Sales'].mean()

segment_avg_sales
```

```python
# 10. Segment Pie Chart
```

```python
segment_counts = customer_segment['Segment'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    segment_counts,
    labels=segment_counts.index,
    autopct='%1.1f%%'
)

plt.title("Customer Segments Share")
plt.savefig("../charts/customers_segments_share.png")
plt.show()
```
